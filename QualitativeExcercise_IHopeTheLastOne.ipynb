```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import ast

# Configurazione grafici
sns.set_theme(style="whitegrid", palette="deep")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
```

```python
df = pd.read_csv('imdb_top_250_movies.csv')

# Visualizziamo le prime righe
display(df.head())

# Dimensioni del dataset
print("Numero di righe:", df.shape[0])
print("Numero di colonne:", df.shape[1])
```

```python
# Statistiche descrittive
display(df.describe(include="all").T)

# Numero di valori mancanti per colonna
missing = df.isnull().sum().sort_values(ascending=False)

display(
    missing[missing > 0].to_frame("Missing values")
)

# Percentuale di valori mancanti
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

display(
    missing_percentage[missing_percentage > 0]
    .to_frame("Missing percentage (%)")
)
```

```python
list_columns = [
    "genres",
    "interests",
    "countriesOfOrigin",
    "spokenLanguages",
    "filmingLocations",
    "productionCompanies",
    "externalLinks"
]

def parse_list(value):
    if pd.isna(value):
        return []

    if isinstance(value, list):
        return value

    try:
        result = ast.literal_eval(value)
        return result if isinstance(result, list) else []
    except (ValueError, SyntaxError):
        return []

for col in list_columns:
    if col in df.columns:
        df[col] = df[col].apply(parse_list)
```

```python
numeric_columns = [
    "startYear",
    "endYear",
    "runtimeMinutes",
    "averageRating",
    "numVotes",
    "metascore",
    "budget",
    "grossWorldwide"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["releaseDate"] = pd.to_datetime(
    df["releaseDate"], errors="coerce"
)

# Creiamo una copia senza colonne monetarie
analysis_df = df.drop(
    columns=["budget", "grossWorldwide"],
    errors="ignore"
)

print("Dataset pronto per l'analisi!")
```

```python
year_counts = (
    analysis_df["startYear"]
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(14, 6))

plt.plot(
    year_counts.index,
    year_counts.values,
    marker="o",
    markersize=3
)

plt.title("Number of Movies Released per Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

```python
analysis_df["decade"] = (
    analysis_df["startYear"] // 10 * 10
)

decade_counts = analysis_df["decade"].value_counts().sort_index()

plt.figure(figsize=(12, 6))

sns.barplot(
    x=decade_counts.index.astype("Int64").astype(str),
    y=decade_counts.values
)

plt.title("Movies per Decade")
plt.xlabel("Decade")
plt.ylabel("Number of Movies")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=analysis_df,
    x="averageRating",
    bins=20,
    kde=True
)

plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

```python
print("Average rating:", analysis_df["averageRating"].mean())
print("Median rating:", analysis_df["averageRating"].median())
print("Standard deviation:", analysis_df["averageRating"].std())

print("\nTop 10 highest-rated movies:")

display(
    analysis_df[
        ["primaryTitle", "startYear", "averageRating", "numVotes"]
    ]
    .sort_values("averageRating", ascending=False)
    .head(10)
)
```

```python
votes_df = analysis_df[
    (analysis_df["numVotes"] > 0) &
    (analysis_df["averageRating"].notna())
].copy()

plt.figure(figsize=(11, 6))

sns.scatterplot(
    data=votes_df,
    x="numVotes",
    y="averageRating",
    alpha=0.4
)

plt.xscale("log")

plt.title("IMDb Rating vs Number of Votes")
plt.xlabel("Number of Votes (log scale)")
plt.ylabel("IMDb Rating")

plt.tight_layout()
plt.show()
```
